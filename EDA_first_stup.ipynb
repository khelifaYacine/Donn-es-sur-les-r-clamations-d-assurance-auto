3. Corrélations
heatmap sur les variables numériques

inspection des variables fortement corrélées

```python
df_corr = df.corr(numeric_only = True)
```

```python
df_corr
```

```python
# Affichage de la heatmap de la matrice de corrélation 
plt.figure(figsize=(10,5))
plt.title("Matrice de corrélation")
sns.heatmap(df_corr, annot=True, cmap="viridis", linewidths=0.5)
plt.show()
```

```python
# 1) Corrélations simples (numérique ↔ binaire) avec Spearman
df["fraud_flag"] = df["fraud_reported"].map({"Y":1, "N":0})

cols = ["total_claim_amount", "injury_claim", "property_claim", "vehicle_claim"]

df[cols + ["fraud_flag"]].corr(method="spearman")
```

```python
#2) Comparaison des distributions (fraude vs non fraude) 
for col in cols:
    plt.figure(figsize=(10,5))
    sns.boxplot(data=df, x="fraud_reported", y=col)
    plt.title(f"{col} selon fraude")
    plt.show()
```

```python
# 3) Test statistique : Mann–Whitney U
from scipy.stats import mannwhitneyu

for col in cols:
    fraud = df[df.fraud_reported=="Y"][col]
    non_fraud = df[df.fraud_reported=="N"][col]
    stat, p = mannwhitneyu(fraud, non_fraud, alternative="two-sided")
    print(col, "p-value =", p)
```

```python
# 4) Visualisation avancée : KDE (densités)
for col in cols:
    plt.figure(figsize=(10,5))
    sns.kdeplot(data=df, x=col, hue="fraud_reported", fill=True)
    plt.title(f"Densité de {col} selon fraude")
    plt.show()
```

#### Synthèse correlaltion 

Les montants des sinistres présentent une relation positive mais faible avec la fraude (corrélations Spearman entre 0.08 et 0.14).  
Les fraudeurs ont tendance à déclarer des montants légèrement plus élevés, ce qui est confirmé par les boxplots et les densités.
Les tests Mann–Whitney montrent que ces différences sont statistiquement significatives, mais la force de la relation reste modérée.
Par ailleurs, les montants sont fortement corrélés entre eux (jusqu’à 0.96), ce qui impose de ne pas utiliser simultanément total_claim_amount et ses sous-composantes dans un modèle.
Conclusion : les montants contribuent à la détection de fraude, mais ne suffisent pas à eux seuls.
