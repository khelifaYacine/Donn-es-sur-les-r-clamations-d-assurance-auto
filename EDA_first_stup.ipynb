2. Analyse bivariée

incident_severity

```python
# 1. fraud_reported vs incident_severity:
plt.figure(figsize=(12,6))
sns.countplot(data=df, x="incident_severity", hue="fraud_reported")
plt.title("Fraude selon la sévérité de l'incident")
plt.xticks(rotation=45)
plt.show()

```

```python
# Tableau de proportions
pd.crosstab(df.incident_severity, df.fraud_reported, normalize='index')
```

```python
# Test statistique : Chi²
from scipy.stats import chi2_contingency
chi2_contingency(pd.crosstab(df.incident_severity, df.fraud_reported))
```

```python
# 2. fraud_reported vs collision_type 
plt.figure(figsize=(10,5))
plt.title("Fraude selon le type de collision")
sns.countplot(data =df, x= "collision_type", hue = "fraud_reported")
plt.xlabel ("Type de collision")
plt.ylabel("nombre de cas")
plt.show()
```

```python
# 3. fraud_reported vs auto_make
plt.figure (figsize =(10,5))
plt.title ("Fraude selon la marque de voiture")
sns.countplot(data = df, x ="auto_make", hue = "fraud_reported")
plt.xlabel ("Marque de voiture")
plt.xticks(rotation = 45)
plt.ylabel ("nombre de cas")
plt.show()
```
